```python
# Modular Calculator Mini- Program


# Asks the user for two numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Asks the user to choose an operation from the options and capitalizes the user's answer
operation_options = ["ADDITION", "SUBTRACTION", "MULTIPLICATION", "DIVISION"]
chosen_operation = input("Choose an operation( ADDITION, SUBTRACTION, MULTIPLICATION, DIVISION ):  ").upper()

# One of the following functions will return a value to what called it in the first place
def add(a, b):
  return a + b
def subtract(a, b):
  return a - b
def multiply(a, b):
  return a * b
def divide(a, b):
  if num2 > 0:
    return a / b

  # This shows an error message when the return value is undefined
  else:
    print("Error! Cannot divide by zero! Please enter a different number!")

# Assigns a value to a variable based on the chosen operation
if chosen_operation == "ADDITION":
  final_answer = add(num1, num2)
elif chosen_operation == "SUBTRACTION" :
  final_answer = subtract(num1, num2)
elif chosen_operation == "MULTIPLICATION":
  final_answer = multiply(num1, num2)
elif chosen_operation == "DIVISION":
  final_answer = divide(num1, num2)

# This shows an error message when the user does not choose from the given options
else:
  print("Invalid Choice! Please choose from the options!")

# This calls the function that was assigned to this variable based on the operation used
print(final_answer)


# Short Reflection:
# 1. The functions that I created in my calculator program are "def add(a,b):" , "def subtract(a, b):", "def multiply(a, b):", "def divide(a, b):"
# 2. The parameters that my functions used were 'a' and 'b'
# 3. The arguments passed when the functions were called were "num1" and "num2"
# 4. My program defined functions and those functions returned values to the variable "final_answer"
# 5. It's better to divide the calculator program into functions instead of writing everything in one block because, It's easier to replicate, other programmers can use your code,
# It's easier to understand, and it makes group project more collaborative
```
